```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()

# Get the uploaded file name
filename = list(uploaded.keys())[0]

# Load the dataset
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
# 11. Check unique values
print("\nUnique Values in Each Column:")
print(df.nunique())
```

```python
# 12. Separate numerical columns
# ============================================

numeric_columns = df.select_dtypes(include=np.number).columns

print("\nNumerical Columns:")
print(numeric_columns.tolist())
```

```python
# 13. Histograms
# ============================================

df[numeric_columns].hist(
    figsize=(15, 12),
    bins=20
)

plt.suptitle("Distribution of Numerical Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# 14. Boxplots - Detect Outliers
# ============================================

plt.figure(figsize=(15, 8))

sns.boxplot(data=df[numeric_columns])

plt.title("Boxplot of Numerical Features")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 15. Correlation Matrix


plt.figure(figsize=(12, 8))

correlation = df[numeric_columns].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python

# 16. Player Performance Score Distribution


plt.figure(figsize=(8, 5))

sns.histplot(
    df["Player_Performance_Score"],
    bins=20,
    kde=True
)

plt.title("Player Performance Score Distribution")
plt.xlabel("Performance Score")
plt.ylabel("Number of Players")
plt.show()
```

```python
# 17. Runs vs Performance Score


plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Runs",
    y="Player_Performance_Score"
)

plt.title("Runs vs Player Performance Score")
plt.xlabel("Runs")
plt.ylabel("Performance Score")
plt.show()
```

```python
# 18. Batting Average vs Performance Score
# ============================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Batting_Average",
    y="Player_Performance_Score"
)

plt.title("Batting Average vs Performance Score")
plt.xlabel("Batting Average")
plt.ylabel("Performance Score")
plt.show()
```

```python
# 19. Strike Rate vs Performance Score
# ============================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Strike_Rate",
    y="Player_Performance_Score"
)

plt.title("Strike Rate vs Player Performance Score")
plt.xlabel("Strike Rate")
plt.ylabel("Performance Score")
plt.show()
```

```python
# ============================================
# 20. Wickets vs Performance Score
# ============================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Wickets",
    y="Player_Performance_Score"
)

plt.title("Wickets vs Player Performance Score")
plt.xlabel("Wickets")
plt.ylabel("Performance Score")
plt.show()
```

```python
# 21. Catches vs Performance Score
# ============================================

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Catches",
    y="Player_Performance_Score"
)

plt.title("Catches vs Player Performance Score")
plt.xlabel("Catches")
plt.ylabel("Performance Score")
plt.show()
```
